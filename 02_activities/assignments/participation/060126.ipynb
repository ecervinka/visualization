## Visualization Examples: 
- Life Expectancy Graph - bad visualization - pretty however, legend is unnecessary, 3D/shadows make it hard to gauge the life expectancy values, bad font choice, accuracy concerns
- Monstrous Costs - 3d format makes it hard to read, lack of labels, interesting to look at
- Aspect Ratios - scale of the aspect ratios distorts the data, one makes it seem as though there is a sharper increase, can be edited to change the message of the author, no source or title or legend - accuracy is a concern
- Presidential Run - total summation is more than 100%, labels are confusing, lack of contrast, proportions are strange, unsure of source
- Wind Map - pleasing to look at, interesting, may be improved with colour because the differences in wind speeds are hard to distinguish, scale is not linear

```python
## Activity & Live Code - Importing Libraries 
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50) #does not inclusde 50, everything starts from 0 in python
y = np.random.randint(0, 100,50)
```

```python
# scatter plot
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y) #importing x and y to create the actual scatterplot
# ax.bar(x,y) - will put the bars right on top if run in same cell
```

```python
# bar graph 
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x,y)
```

```python
# line plot
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
```

```python
# histogram - remember u only need one variable, in this case, y
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y)
```

```python
# adding labels or titles
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)

#adding the labels
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel("Years since start")
fig.tight_layout()
```

```python
# modifying lables and titles
font1 = {'family':'sans-serif','color':'blue','size':20} #automatically recognize as dic oobject with {}
font2 = {'family':'monospace','color':'green','size':14}
         
```

```python
#adding fontdict arguments
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y)

#adding the labels
ax.set_title('Total growth over time', fontdict = font1) #loc = "left")
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel("Years since start", fontdict = font2)
fig.tight_layout()
```

```python
#modifying data points
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x,y,
           marker = '*',
           color = 'indigo')

fig.show() #used by python to display the figure, outside of jupyter
```

```python
#modifying data points
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y,
           marker = '*',
           color = 'indigo',
           linestyle = '--', #other linestyles available
           linewidth =1) #need to use lineplot for this 

fig.show()
```

```python
#modifying colour - check online library, or use hexcode

fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y,
           marker = '*',
           color = '#7425b9',
           linestyle = '--', #other linestyles available
           linewidth =1) #need to use lineplot for this 

fig.show()
```

```python
#modifying colour - the stars
fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y,
           marker = '*',
           markersize=12,
           color = '#7425b9',
           linestyle = '--', #other linestyles available
           linewidth =2,
           markeredgecolor = '#fa9359', 
           markerfacecolor= '#000000') #need to use lineplot for this 

```

```python
#adding grid lines
fig, ax = plt.subplots(figsize=(5,3))

ax.plot(x,y,
           marker = '*',
           markersize=12,
           color = '#7425b9',
           linestyle = '--', #other linestyles available
           linewidth =2,
           markeredgecolor = '#fa9359', 
           markerfacecolor= '#000000') #need to use lineplot for this )

ax.grid(axis = 'y')

ax.grid(axis = 'y')
ax.grid (axis= 'y', color = 'blue', linewidth = 2, linestyle = '-.') #adds to existing plot
```

### Comments regarding the quality of the data viz 
- aesthetic - the correlegram is not very complex in it's representation. Each plot is limited to one colour, and the labels only include variable names, with no titles either. While it is simple and clean in its layout, some adjustments with labels and colour would make it clearer. 
- substantive - the visualization is honest and accurate. The simple layout is clean and shows the correlations between different variables with no colours or language that evokes emotions or communicates bias. However, the plot could use a title and source to make it more clear and trustworthy. 
- perceptual - yes, it is clear that the goal is to provide a visualization that allows the viewer to look at the relationship between different variables, however, with the limited information provided on the graph, the meaning of the exact variables are unclear.
