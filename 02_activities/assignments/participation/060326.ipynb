## Data viz as rhetorical objects
- rhetoric is the act of communicating effectively and persuasively
- we make choices in our representation
- can be factual, without being neutral 
- three important qualities: aesthetic, substantive, perceptual 

### Purposes of data viz
- persuading, comparison, evaluating, exploring 

### Intended audience
- age, education, expertise, accessibility

### Intended medium
- print, web, poster, presentation

### Dataset types
- tables, networks, fields, geometries - four basic types
- also have multidimensional table, trees

**knowing the attributes of the data helps us determine how to visualize - items, attributes, links, positions, grids 

### Effective Visualization 
- depends on purpose, audience, medium
- known space, consideration space, proposal space, selected solution 

## Cognitive Psychology 
- can take advantage of human cognition to help understand data more intuitively
- Gestalt Principles - set of cognitive theories for how visual info is organized, used in UX/viz
    - proximity, similarity, continuity, closure, enclosure, connection 


### Cognitive Load
- need to consider cognitive load: intrinsic, germane, extraneous** most in our control
    - familiar vs rare chart types
    - accurate vs approximate interpretation
    - concise vs detailed composition
    - explanatory vs exploratory composition 

**cognition over memory - working memory is limite, better to show side by side

EX) 3D increases cognitive load - 3D better for data that is 3D in nature

### Perceived factual bias
- four conventions of data viz that reinforces objectivity and factual basis
    - 2D image
    - clean layouts
    - geometric shapes and lines
    - sources 

### Provenance rhetoric
- citing the source - inclusion increases persuasiveness
- transparency and trustworthiness
